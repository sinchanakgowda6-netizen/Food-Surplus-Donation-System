```python
def calculate_distance(food, ngo):
    donor_location = (food["lat"], food["lon"])
    ngo_location = (ngo["lat"], ngo["lon"])

    return geodesic(donor_location, ngo_location).km


def match_food_to_ngos(food, ngos):
    matches = []

    for _, ngo in ngos.iterrows():

        # Food category compatibility
        if food["category"] not in ngo["accepts"]:
            continue

        # Storage requirement compatibility
        if food["storage"] not in ngo["storage"]:
            continue

        # Recipient capacity check
        if ngo["capacity_kg"] <= 0:
            continue

        # Remaining usable pickup time
        if food["pickup_hours"] <= 0:
            continue

        distance = calculate_distance(food, ngo)

        # Estimate transport time: 20 km/h for demo purposes
        travel_hours = distance / 20

        # Do not recommend if estimated travel exceeds pickup window
        if travel_hours >= food["pickup_hours"]:
            continue

        # Normalize component scores to 0-100
        distance_score = max(0, 100 - distance * 5)

        capacity_score = min(
            100,
            ngo["capacity_kg"] / food["quantity_kg"] * 100
        )

        urgency_score = max(
            0,
            100 - (food["expiry_hours"] / 48) * 100
        )

        # Weighted recommendation score
        final_score = (
            0.40 * distance_score +
            0.35 * capacity_score +
            0.25 * urgency_score
        )

        matches.append({
            "Food ID": food["food_id"],
            "Donor": food["donor"],
            "Food": food["food"],
            "NGO": ngo["ngo"],
            "Distance (km)": round(distance, 2),
            "Capacity (kg)": ngo["capacity_kg"],
            "Food quantity (kg)": food["quantity_kg"],
            "Match score": round(final_score, 2),
            "Estimated travel (hours)": round(travel_hours, 2)
        })

    return sorted(
        matches,
        key=lambda item: item["Match score"],
        reverse=True
    )


# Test the matching engine
test_food = food_surplus.iloc[0]
results = match_food_to_ngos(test_food, ngos)

if results:
    display(pd.DataFrame(results))
else:
    print("No compatible recipient found.")
```

```python
def run_surplus_agent(food_df, ngo_df):
    decisions = []

    for _, food in food_df.iterrows():

        # Basic eligibility checks
        if not bool(food["safe"]):
            decisions.append({
                "Food ID": food["food_id"],
                "Donor": food["donor"],
                "Status": "HOLD - Safety review required",
                "Recommended NGO": "None",
                "Match score": 0
            })
            continue

        if food["quantity_kg"] <= 0 or food["expiry_hours"] <= 0:
            decisions.append({
                "Food ID": food["food_id"],
                "Donor": food["donor"],
                "Status": "HOLD - Invalid quantity or timing",
                "Recommended NGO": "None",
                "Match score": 0
            })
            continue

        # Find compatible recipient organizations
        ranked_matches = match_food_to_ngos(food, ngo_df)

        if not ranked_matches:
            decisions.append({
                "Food ID": food["food_id"],
                "Donor": food["donor"],
                "Status": "NO MATCH - Manual review",
                "Recommended NGO": "None",
                "Match score": 0
            })
            continue

        best_match = ranked_matches[0]

        decisions.append({
            "Food ID": food["food_id"],
            "Donor": food["donor"],
            "Status": "AWAITING HUMAN APPROVAL",
            "Recommended NGO": best_match["NGO"],
            "Distance (km)": best_match["Distance (km)"],
            "Match score": best_match["Match score"],
            "Quantity (kg)": food["quantity_kg"]
        })

    return pd.DataFrame(decisions)


agent_results = run_surplus_agent(food_surplus, ngos)

print("SurplusSync AI agent decisions:")
display(agent_results)
```

```python
import gradio as gr

def analyze_surplus(food_name, category, quantity, expiry, pickup,
                    storage, donor_lat, donor_lon):

    new_food = {
        "food_id": "LIVE001",
        "donor": "New Donation",
        "food": food_name,
        "category": category,
        "quantity_kg": float(quantity),
        "expiry_hours": float(expiry),
        "lat": float(donor_lat),
        "lon": float(donor_lon),
        "pickup_hours": float(pickup),
        "storage": storage,
        "safe": True
    }

    if (not food_name.strip()
        or quantity <= 0
        or expiry <= 0
        or pickup <= 0
        or not (-90 <= donor_lat <= 90)
        or not (-180 <= donor_lon <= 180)):
        return "Please enter valid food, quantity, timing, and coordinates.", pd.DataFrame()

    matches = match_food_to_ngos(new_food, ngos)

    if not matches:
        return (
            "No compatible recipient found. Manual review is required.",
            pd.DataFrame()
        )

    output = pd.DataFrame(matches)

    best = matches[0]

    summary = (
        f"Top recommendation: {best['NGO']}\n"
        f"Distance: {best['Distance (km)']} km\n"
        f"Match score: {best['Match score']}/100\n"
        f"Estimated travel: {best['Estimated travel (hours)']} hours\n\n"
        "Status: Recommendation only — verify food safety, "
        "recipient acceptance, and delivery before confirming."
    )

    return summary, output


with gr.Blocks(title="SurplusSync AI") as demo:

    gr.Markdown("# SurplusSync AI")
    gr.Markdown(
        "### Autonomous Food Surplus Redistribution & Waste Reduction"
    )

    with gr.Row():
        food_name = gr.Textbox(
            label="Food name",
            value="Fresh vegetables"
        )
        category = gr.Dropdown(
            choices=["prepared", "produce", "bakery"],
            value="produce",
            label="Food category"
        )

    with gr.Row():
        quantity = gr.Number(label="Quantity (kg)", value=20)
        expiry = gr.Number(label="Hours until use-by/expiry", value=24)
        pickup = gr.Number(label="Pickup window (hours)", value=4)

    storage = gr.Dropdown(
        choices=["hot", "ambient", "chilled"],
        value="ambient",
        label="Storage requirement"
    )

    with gr.Row():
        donor_lat = gr.Number(label="Donor latitude", value=12.9716)
        donor_lon = gr.Number(label="Donor longitude", value=77.5946)

    run_button = gr.Button("Find Best NGO Matches", variant="primary")

    summary = gr.Textbox(label="Agent recommendation", lines=5)
    table = gr.Dataframe(label="Ranked recipient matches")

    run_button.click(
        fn=analyze_surplus,
        inputs=[
            food_name, category, quantity, expiry, pickup,
            storage, donor_lat, donor_lon
        ],
        outputs=[summary, table]
    )

demo.launch()
```

```python
import gradio as gr
import plotly.express as px
import pandas as pd
import os

# Use the datasets already created in Steps 2A and 2B
dashboard_food = food_surplus.copy()
dashboard_ngos = ngos.copy()


def get_dashboard_data(category_filter="All categories"):
    """Generate dashboard statistics, charts, and tables."""

    if category_filter == "All categories":
        food_view = dashboard_food.copy()
    else:
        food_view = dashboard_food[
            dashboard_food["category"] == category_filter
        ].copy()

    # Recalculate recommendations from the current dataset
    decisions = run_surplus_agent(dashboard_food, dashboard_ngos)

    if category_filter != "All categories":
        visible_ids = food_view["food_id"].tolist()
        decisions = decisions[
            decisions["Food ID"].isin(visible_ids)
        ].copy()

    total_kg = food_view["quantity_kg"].sum()
    total_listings = len(food_view)

    matched_ids = decisions.loc[
        decisions["Status"] == "AWAITING HUMAN APPROVAL",
        "Food ID"
    ].tolist()

    matched_kg = food_view.loc[
        food_view["food_id"].isin(matched_ids),
        "quantity_kg"
    ].sum()

    meals = int(matched_kg / 0.4)  # Illustrative estimate only
    match_rate = (
        len(matched_ids) / total_listings * 100
        if total_listings else 0
    )

    # Category chart
    category_data = (
        food_view.groupby("category", as_index=False)["quantity_kg"]
        .sum()
    )

    fig_category = px.bar(
        category_data,
        x="category",
        y="quantity_kg",
        color="category",
        title="Food Surplus by Category",
        labels={
            "category": "Food Category",
            "quantity_kg": "Quantity (kg)"
        },
        template="plotly_white"
    )

    # NGO chart: quantity associated with recommendations
    if not decisions.empty and "Recommended NGO" in decisions.columns:
        accepted_for_review = decisions[
            decisions["Status"] == "AWAITING HUMAN APPROVAL"
        ].copy()

        ngo_data = (
            accepted_for_review.groupby(
                "Recommended NGO", as_index=False
            )["Quantity (kg)"].sum()
            if not accepted_for_review.empty
            else pd.DataFrame(
                columns=["Recommended NGO", "Quantity (kg)"]
            )
        )
    else:
        ngo_data = pd.DataFrame(
            columns=["Recommended NGO", "Quantity (kg)"]
        )

    fig_ngo = px.bar(
        ngo_data,
        x="Recommended NGO",
        y="Quantity (kg)",
        title="Food Quantity by Recommended NGO",
        labels={
            "Recommended NGO": "Recipient",
            "Quantity (kg)": "Recommended Quantity (kg)"
        },
        template="plotly_white"
    )

    metrics = f"""
### 📊 SurplusSync Impact Overview

| Metric | Current value |
|---|---:|
| Food listings | {total_listings} |
| Total food listed | {total_kg:.1f} kg |
| Food with a recommended match | {matched_kg:.1f} kg |
| Listings with recommendations | {match_rate:.1f}% |
| Estimated meal equivalents | {meals} |
| Registered NGO partners | {len(dashboard_ngos)} |

*Recommendations are not confirmed donations. Meal equivalents use an
illustrative conversion of 0.4 kg per meal.*
"""

    # Ensure tables always have usable columns
    table_columns = [
        "food_id", "donor", "food", "category",
        "quantity_kg", "expiry_hours", "pickup_hours"
    ]

    food_table = food_view[table_columns].copy()

    return metrics, fig_category, fig_ngo, food_table, decisions


def export_dashboard(category_filter):
    """Export the currently selected food listings."""
    if category_filter == "All categories":
        export_data = dashboard_food.copy()
    else:
        export_data = dashboard_food[
            dashboard_food["category"] == category_filter
        ].copy()

    path = "/content/surplussync_dashboard_export.csv"
    export_data.to_csv(path, index=False)
    return path


# Build the interactive interface
with gr.Blocks(
    theme=gr.themes.Soft(),
    title="SurplusSync AI Dashboard"
) as dashboard:

    gr.Markdown("""
    # 🌱 SurplusSync AI
    ### Food Surplus Redistribution & Waste Reduction

    Monitor donations, explore food categories, and inspect AI-generated
    recipient recommendations.
    """)

    with gr.Row():
        category_filter = gr.Dropdown(
            choices=[
                "All categories", "prepared", "produce", "bakery"
            ],
            value="All categories",
            label="Filter by food category"
        )

        refresh_button = gr.Button(
            "🔄 Refresh Dashboard",
            variant="primary"
        )

    metrics_output = gr.Markdown()

    with gr.Row():
        category_chart = gr.Plot(label="Food Category Analysis")
        ngo_chart = gr.Plot(label="NGO Recommendations")

    gr.Markdown("## 🍲 Food Donation Inventory")

    food_table_output = gr.Dataframe(
        label="Food listings",
        interactive=False
    )

    gr.Markdown("## 🤖 AI Matching Decisions")

    decisions_output = gr.Dataframe(
        label="Recipient recommendations",
        interactive=False
    )

    with gr.Row():
        export_button = gr.Button("📥 Export Food Data to CSV")
        download_file = gr.File(label="Download CSV")

    # Load the initial dashboard
    dashboard.load(
        fn=get_dashboard_data,
        inputs=category_filter,
        outputs=[
            metrics_output,
            category_chart,
            ngo_chart,
            food_table_output,
            decisions_output
        ]
    )

    # Update when the category changes
    category_filter.change(
        fn=get_dashboard_data,
        inputs=category_filter,
        outputs=[
            metrics_output,
            category_chart,
            ngo_chart,
            food_table_output,
            decisions_output
        ]
    )

    # Refresh manually
    refresh_button.click(
        fn=get_dashboard_data,
        inputs=category_filter,
        outputs=[
            metrics_output,
            category_chart,
            ngo_chart,
            food_table_output,
            decisions_output
        ]
    )

    # Export filtered records
    export_button.click(
        fn=export_dashboard,
        inputs=category_filter,
        outputs=download_file
    )

dashboard.launch()
```

```python
import gradio as gr
from datetime import datetime, timezone
import pandas as pd
import os

def submit_donation(
    donor, food_name, category, quantity,
    expiry, pickup, storage, latitude, longitude,
    safety_confirmed
):
    global donations_db

    # Validate inputs
    if not donor.strip() or not food_name.strip():
        return "Enter the donor name and food name.", donations_db

    if quantity <= 0 or expiry <= 0 or pickup <= 0:
        return "Quantity and timing values must be positive.", donations_db

    if not (-90 <= latitude <= 90 and -180 <= longitude <= 180):
        return "Please enter valid latitude and longitude.", donations_db

    # Require an explicit safety confirmation
    if not safety_confirmed:
        return (
            "Donation not submitted. Verify food safety and applicable "
            "handling requirements before proceeding.",
            donations_db
        )

    # Generate a unique donation ID
    food_id = "SS-" + datetime.now(timezone.utc).strftime("%Y%m%d%H%M%S%f")

    food = {
        "food_id": food_id,
        "donor": donor.strip(),
        "food": food_name.strip(),
        "category": category,
        "quantity_kg": float(quantity),
        "expiry_hours": float(expiry),
        "lat": float(latitude),
        "lon": float(longitude),
        "pickup_hours": float(pickup),
        "storage": storage,
        "safe": True
    }

    # Find compatible NGOs
    matches = match_food_to_ngos(food, ngos)

    if matches:
        best = matches[0]
        recommended_ngo = best["NGO"]
        distance_km = best["Distance (km)"]
        score = best["Match score"]
        status = "AWAITING HUMAN APPROVAL"
    else:
        recommended_ngo = "None"
        distance_km = None
        score = None
        status = "NO MATCH - MANUAL REVIEW"

    new_record = {
        **food,
        "recommended_ngo": recommended_ngo,
        "distance_km": distance_km,
        "match_score": score,
        "status": status,
        "created_at": datetime.now(timezone.utc).isoformat()
    }

    # Save to persistent Google Drive CSV
    new_row = pd.DataFrame([new_record], columns=DONATION_COLUMNS)

    donations_db = pd.concat(
        [donations_db, new_row],
        ignore_index=True
    )

    donations_db.to_csv(DONATION_FILE, index=False)

    message = (
        f"Donation saved successfully!\n"
        f"Donation ID: {food_id}\n"
        f"Recommended NGO: {recommended_ngo}\n"
        f"Match score: {score if score is not None else 'Unavailable'}\n"
        f"Status: {status}\n\n"
        "A recommendation is not a confirmed donation. "
        "Obtain recipient acceptance and verify food safety before delivery."
    )

    return message, donations_db


with gr.Blocks(title="SurplusSync AI - Live Donations") as donation_app:
    gr.Markdown("# 🌱 SurplusSync AI")
    gr.Markdown("## Register a new surplus food donation")

    with gr.Row():
        donor = gr.Textbox(label="Donor / Restaurant Name")
        food_name = gr.Textbox(label="Food Name")

    with gr.Row():
        category = gr.Dropdown(
            ["prepared", "produce", "bakery"],
            value="produce",
            label="Food Category"
        )
        storage = gr.Dropdown(
            ["hot", "ambient", "chilled"],
            value="ambient",
            label="Storage Requirement"
        )

    with gr.Row():
        quantity = gr.Number(label="Quantity (kg)", value=10)
        expiry = gr.Number(label="Hours until use-by / expiry", value=24)
        pickup = gr.Number(label="Pickup Window (hours)", value=4)

    gr.Markdown("### Pickup Location")
    with gr.Row():
        latitude = gr.Number(label="Latitude", value=12.9716)
        longitude = gr.Number(label="Longitude", value=77.5946)

    safety_confirmed = gr.Checkbox(
        label="I have verified food safety and appropriate storage/handling",
        value=False
    )

    submit_button = gr.Button("Save Donation & Find NGO", variant="primary")
    result = gr.Textbox(label="Agent Recommendation", lines=6)
    records_table = gr.Dataframe(
        label="Saved Donation Records",
        value=donations_db,
        interactive=False
    )

    submit_button.click(
        fn=submit_donation,
        inputs=[
            donor, food_name, category, quantity,
            expiry, pickup, storage, latitude, longitude,
            safety_confirmed
        ],
        outputs=[result, records_table]
    )

donation_app.launch(share=True, debug=True)
```

```python
# STEP 11B: NGO acceptance and rejection system

import pandas as pd
from datetime import datetime, timezone
import uuid
import os

# Check that Step 11A has been completed
if "COORDINATION_FILE" not in globals():
    raise RuntimeError("Please run Step 11A first.")

if "donations_db" not in globals():
    donations_db = pd.read_csv(DONATION_FILE)

if "coordination_db" not in globals():
    coordination_db = pd.read_csv(COORDINATION_FILE)


def create_ngo_offer(food_id, ngo_name):
    """Create an offer for an NGO."""

    global coordination_db

    if food_id not in donations_db["food_id"].astype(str).values:
        print("Error: Donation ID not found.")
        return

    offer_id = "OFFER-" + uuid.uuid4().hex[:8].upper()

    new_offer = {
        "offer_id": offer_id,
        "food_id": food_id,
        "ngo_name": ngo_name,
        "status": "AWAITING RESPONSE",
        "offered_at": datetime.now(timezone.utc).isoformat(),
        "responded_at": "",
        "collection_status": "NOT COLLECTED"
    }

    coordination_db = pd.concat(
        [coordination_db, pd.DataFrame([new_offer])],
        ignore_index=True
    )

    coordination_db.to_csv(COORDINATION_FILE, index=False)

    print("NGO offer created successfully!")
    print("Offer ID:", offer_id)
    print("Donation ID:", food_id)
    print("NGO:", ngo_name)
    print("Status: AWAITING RESPONSE")

    return offer_id


def respond_to_ngo_offer(offer_id, response):
    """Record an NGO's acceptance or rejection."""

    global coordination_db

    response = response.strip().upper()

    if response not in ["ACCEPTED", "REJECTED"]:
        print("Invalid response. Use ACCEPTED or REJECTED.")
        return

    matching_rows = coordination_db.index[
        coordination_db["offer_id"].astype(str) == str(offer_id)
    ].tolist()

    if not matching_rows:
        print("Error: Offer ID not found.")
        return

    idx = matching_rows[0]

    if coordination_db.at[idx, "status"] != "AWAITING RESPONSE":
        print("This offer has already received a response.")
        return

    coordination_db.at[idx, "status"] = response
    coordination_db.at[idx, "responded_at"] = (
        datetime.now(timezone.utc).isoformat()
    )

    coordination_db.to_csv(COORDINATION_FILE, index=False)

    print("NGO response saved successfully!")
    print("Offer ID:", offer_id)
    print("Response:", response)

    if response == "REJECTED":
        print("Next action: Find another eligible NGO.")
    else:
        print("Next action: Arrange collection and await confirmation.")


print("STEP 11B READY!")
print("Functions created: create_ngo_offer() and respond_to_ngo_offer()")
```

```python
# ================================================================
# FOOD SURPLUS DONATION SYSTEM
# FINAL WORKFLOW, NGO MATCHING, OFFER CREATION & REASSIGNMENT
# ================================================================

import pandas as pd
import uuid
from datetime import datetime

print("=" * 65)
print(" FOOD SURPLUS DONATION SYSTEM - FINAL WORKFLOW")
print("=" * 65)

# ------------------------------------------------
# STEP 1: CHECK REQUIRED DATABASES
# ------------------------------------------------

required_variables = [
    "donations_db",
    "ngos",
    "coordination_db"
]

missing = [
    name for name in required_variables
    if name not in globals()
]

if missing:
    print("ERROR: Required database variables are missing:", missing)
    print("Run the earlier notebook cells that load these databases.")
else:
    # Work with copies so the original tables are not changed
    # until the workflow explicitly saves an update.
    donations = donations_db.copy()
    ngo_table = ngos.copy()
    coordination = coordination_db.copy()

    print("\nSTEP 1: Database checks")

    print("Donation database:", len(donations), "records")
    print("NGO database:", len(ngo_table), "records")
    print("Coordination database:", len(coordination), "records")

    # ------------------------------------------------
    # STEP 2: CHECK REQUIRED COLUMNS
    # ------------------------------------------------

    print("\nSTEP 2: Checking database columns")

    required_donation_columns = [
        "food_id", "food", "category",
        "quantity_kg", "expiry_hours"
    ]

    required_ngo_columns = [
        "ngo", "capacity_kg", "accepts", "storage"
    ]

    required_coordination_columns = [
        "offer_id", "food_id", "ngo_name", "status"
    ]

    missing_donation = [
        c for c in required_donation_columns
        if c not in donations.columns
    ]

    missing_ngo = [
        c for c in required_ngo_columns
        if c not in ngo_table.columns
    ]

    missing_coordination = [
        c for c in required_coordination_columns
        if c not in coordination.columns
    ]

    if missing_donation or missing_ngo or missing_coordination:
        print("Cannot continue: required columns are missing.")
        print("Donation columns missing:", missing_donation)
        print("NGO columns missing:", missing_ngo)
        print("Coordination columns missing:", missing_coordination)

    else:
        # ------------------------------------------------
        # STEP 3: NORMALIZE DATA
        # ------------------------------------------------

        print("\nSTEP 3: Preparing data")

        # Use the existing pickup_hours field if available.
        # Otherwise, use expiry_hours as a fallback estimate.
        if "pickup_hours" not in donations.columns:
            donations["pickup_hours"] = donations["expiry_hours"]

        donations["quantity_kg"] = pd.to_numeric(
            donations["quantity_kg"], errors="coerce"
        )

        donations["expiry_hours"] = pd.to_numeric(
            donations["expiry_hours"], errors="coerce"
        )

        donations["pickup_hours"] = pd.to_numeric(
            donations["pickup_hours"], errors="coerce"
        )

        ngo_table["capacity_kg"] = pd.to_numeric(
            ngo_table["capacity_kg"], errors="coerce"
        )

        # Convert list-like or string fields to searchable text.
        def normalized_text(value):
            if isinstance(value, (list, tuple, set)):
                return " ".join(str(x) for x in value).lower()
            return str(value).lower()

        # ------------------------------------------------
        # STEP 4: MATCH FOOD TO ELIGIBLE NGOS
        # ------------------------------------------------

        def match_food_to_ngos(food_row, ngo_data):
            matches = []

            food_category = normalized_text(
                food_row.get("category", "")
            )

            quantity = food_row.get("quantity_kg", 0)
            expiry = food_row.get("expiry_hours", 0)
            pickup = food_row.get("pickup_hours", expiry)

            if pd.isna(quantity) or pd.isna(expiry):
                return pd.DataFrame()

            if pd.isna(pickup) or pickup <= 0 or expiry <= 0:
                return pd.DataFrame()

            for _, ngo in ngo_data.iterrows():
                capacity = ngo.get("capacity_kg", 0)

                if pd.isna(capacity) or capacity < quantity:
                    continue

                accepted_categories = normalized_text(
                    ngo.get("accepts", "")
                )

                # Match the donation category against NGO preferences.
                if food_category not in accepted_categories:
                    continue

                matches.append(ngo.to_dict())

            return pd.DataFrame(matches)

        # ------------------------------------------------
        # STEP 5: CREATE AN NGO OFFER
        # ------------------------------------------------

        def create_ngo_offer(food_row, ngo_name):
            nonlocal_placeholder = None

            new_offer = {
                "offer_id": uuid.uuid4().hex[:8].upper(),
                "food_id": str(food_row["food_id"]),
                "ngo_name": str(ngo_name),
                "status": "AWAITING RESPONSE",
                "offered_at": datetime.now().isoformat()
            }

            # Add optional columns that already exist in the table.
            for col in coordination.columns:
                if col not in new_offer:
                    new_offer[col] = None

            return new_offer

        # ------------------------------------------------
        # STEP 6: FIND AN ALTERNATIVE NGO
        # ------------------------------------------------

        def find_next_ngo(food_id, rejected_ngo):
            food_rows = donations[
                donations["food_id"].astype(str) == str(food_id)
            ]

            if food_rows.empty:
                print("Food donation not found:", food_id)
                return None

            food_row = food_rows.iloc[0]

            candidates = match_food_to_ngos(food_row, ngo_table)

            if candidates.empty:
                print("No eligible NGOs found for this donation.")
                return None

            already_offered = set(
                coordination.loc[
                    coordination["food_id"].astype(str) == str(food_id),
                    "ngo_name"
                ].astype(str)
            )

            already_offered.add(str(rejected_ngo))

            for _, candidate in candidates.iterrows():
                candidate_name = str(candidate["ngo"])

                if candidate_name not in already_offered:
                    offer = create_ngo_offer(
                        food_row, candidate_name
                    )

                    updated_coordination = pd.concat(
                        [
                            coordination,
                            pd.DataFrame([offer])
                        ],
                        ignore_index=True
                    )

                    print("Alternative NGO selected:", candidate_name)
                    print("New offer ID:", offer["offer_id"])

                    return {
                        "offer": offer,
                        "coordination_db": updated_coordination
                    }

            print("No new eligible NGO is available.")
            return None

        # ------------------------------------------------
        # STEP 7: SHOW ELIGIBLE NGOS FOR THE LATEST DONATION
        # ------------------------------------------------

        print("\nSTEP 7: NGO matching")

        if donations.empty:
            print("No donations are currently available.")

        elif ngo_table.empty:
            print("No NGOs are currently registered.")

        else:
            latest_food = donations.iloc[-1]

            print("\nLatest donation:")
            print(
                latest_food[
                    [
                        "food_id", "food", "category",
                        "quantity_kg", "expiry_hours"
                    ]
                ].to_string()
            )

            eligible_ngos = match_food_to_ngos(
                latest_food, ngo_table
            )

            if eligible_ngos.empty:
                print("\nNo eligible NGO matches this donation.")
            else:
                print("\nEligible NGOs:")
                display(eligible_ngos)

                # Create an offer only if one does not already exist
                food_id = str(latest_food["food_id"])

                existing = coordination[
                    coordination["food_id"].astype(str) == food_id
                ]

                if existing.empty:
                    first_ngo = eligible_ngos.iloc[0]["ngo"]

                    offer = create_ngo_offer(
                        latest_food, first_ngo
                    )

                    coordination = pd.concat(
                        [
                            coordination,
                            pd.DataFrame([offer])
                        ],
                        ignore_index=True
                    )

                    print("\nNew offer created.")
                    print("Offer ID:", offer["offer_id"])
                    print("NGO:", first_ngo)
                    print("Status:", offer["status"])

                else:
                    print(
                        "\nAn offer already exists for this donation."
                    )
                    print(existing.to_string(index=False))

        # ------------------------------------------------
        # STEP 8: FINAL VALIDATION
        # ------------------------------------------------

        print("\n" + "=" * 65)
        print("STEP 8: FINAL VALIDATION")
        print("=" * 65)

        print("\nDatabase checks:")

        print(
            "donations_db:",
            "PASS" if not donations.empty else "EMPTY"
        )

        print(
            "ngos:",
            "PASS" if not ngo_table.empty else "EMPTY"
        )

        print(
            "coordination_db:",
            "PASS",
            "- required columns are present"
        )

        print("\nFunction checks:")

        print(
            "match_food_to_ngos:",
            "PASS - function defined"
        )

        print(
            "create_ngo_offer:",
            "PASS - function defined"
        )

        print(
            "find_next_ngo:",
            "PASS - function defined"
        )

        # Update the notebook variables with the current tables.
        donations_db = donations
        coordination_db = coordination

        print("\nCurrent coordination database:")
        display(coordination_db.tail(10))

        print("\n" + "=" * 65)
        print("FINAL WORKFLOW CELL COMPLETED")
        print("=" * 65)

        print(
            "\nNote: This is an in-memory demonstration. "
            "It does not contact NGOs or send real notifications."
        )
```

```python
# ================================================================
# FOOD SURPLUS DONATION SYSTEM
# COMPLETE PROJECT - GITHUB SUBMISSION VERSION
# ================================================================

import pandas as pd
import uuid
import os
from datetime import datetime, timezone

print("=" * 65)
print("       FOOD SURPLUS DONATION MANAGEMENT SYSTEM")
print("=" * 65)

# ================================================================
# STEP 1: CREATE / LOAD DATABASES
# ================================================================

if "donations_db" not in globals():
    donations_db = pd.DataFrame([
        {
            "food_id": "FOOD001",
            "donor": "Green Leaf Restaurant",
            "food": "Fresh vegetables",
            "category": "produce",
            "quantity_kg": 10.0,
            "expiry_hours": 24.0,
            "pickup_hours": 12.0,
            "lat": 12.9716,
            "lon": 77.5946
        }
    ])

if "ngos" not in globals():
    ngos = pd.DataFrame([
        {
            "ngo": "Hope Food Foundation",
            "capacity_kg": 100,
            "lat": 12.975,
            "lon": 77.600,
            "accepts": ["prepared", "produce", "bakery"],
            "storage": ["hot", "ambient", "chilled"]
        },
        {
            "ngo": "Community Care Center",
            "capacity_kg": 40,
            "lat": 12.940,
            "lon": 77.620,
            "accepts": ["produce", "bakery"],
            "storage": ["ambient", "chilled"]
        },
        {
            "ngo": "City Relief Kitchen",
            "capacity_kg": 60,
            "lat": 12.990,
            "lon": 77.570,
            "accepts": ["prepared", "bakery"],
            "storage": ["hot", "ambient", "chilled"]
        },
        {
            "ngo": "Neighborhood Food Bank",
            "capacity_kg": 25,
            "lat": 12.965,
            "lon": 77.605,
            "accepts": ["produce", "prepared", "bakery"],
            "storage": ["hot", "ambient", "chilled"]
        }
    ])

if "coordination_db" not in globals():
    coordination_db = pd.DataFrame(columns=[
        "offer_id",
        "food_id",
        "ngo_name",
        "status",
        "offered_at",
        "responded_at",
        "collection_status"
    ])

# ================================================================
# STEP 2: HELPER FUNCTIONS
# ================================================================

def normalize_list(value):
    """Convert category fields into a consistent list."""
    if isinstance(value, (list, tuple, set)):
        return [str(x).strip().lower() for x in value]

    if pd.isna(value):
        return []

    text = str(value).strip()

    if text.startswith("[") and text.endswith("]"):
        text = text[1:-1]

    return [
        item.strip().strip("'\"").lower()
        for item in text.split(",")
        if item.strip()
    ]


def get_food(food_id):
    """Find a donation by its unique ID."""
    rows = donations_db[
        donations_db["food_id"].astype(str) == str(food_id)
    ]

    if rows.empty:
        return None

    return rows.iloc[0]


def match_food_to_ngos(food_id):
    """Return NGOs that can accept the donation."""
    food = get_food(food_id)

    if food is None:
        return pd.DataFrame()

    quantity = pd.to_numeric(
        food.get("quantity_kg"), errors="coerce"
    )

    expiry = pd.to_numeric(
        food.get("expiry_hours"), errors="coerce"
    )

    pickup = pd.to_numeric(
        food.get("pickup_hours", expiry), errors="coerce"
    )

    if (
        pd.isna(quantity)
        or pd.isna(expiry)
        or pd.isna(pickup)
        or quantity <= 0
        or expiry <= 0
        or pickup <= 0
    ):
        return pd.DataFrame()

    category = str(food["category"]).strip().lower()
    matches = []

    for _, ngo in ngos.iterrows():
        capacity = pd.to_numeric(
            ngo.get("capacity_kg"), errors="coerce"
        )

        if pd.isna(capacity) or capacity < quantity:
            continue

        accepted = normalize_list(ngo.get("accepts", []))

        if category not in accepted:
            continue

        matches.append(ngo.to_dict())

    return pd.DataFrame(matches)


def create_ngo_offer(food_id, ngo_name):
    """Create an offer for an eligible NGO."""
    food = get_food(food_id)

    if food is None:
        print("ERROR: Donation not found.")
        return None

    existing = coordination_db[
        (coordination_db["food_id"].astype(str) == str(food_id))
        & (coordination_db["ngo_name"].astype(str) == str(ngo_name))
    ]

    if not existing.empty:
        print("This NGO already has an offer for this donation.")
        return None

    eligible = match_food_to_ngos(food_id)

    if eligible.empty or str(ngo_name) not in eligible["ngo"].astype(str).tolist():
        print("ERROR: This NGO is not eligible for this donation.")
        return None

    offer = {
        "offer_id": "OFFER-" + uuid.uuid4().hex[:8].upper(),
        "food_id": str(food_id),
        "ngo_name": str(ngo_name),
        "status": "PENDING",
        "offered_at": datetime.now(timezone.utc).isoformat(),
        "responded_at": None,
        "collection_status": "NOT COLLECTED"
    }

    coordination_db.loc[len(coordination_db)] = offer

    print("Offer created successfully.")
    print("Offer ID:", offer["offer_id"])
    print("Food ID:", food_id)
    print("NGO:", ngo_name)
    print("Status:", offer["status"])

    return offer


def respond_to_ngo_offer(offer_id, response):
    """Record an NGO's acceptance or rejection."""
    response = str(response).strip().upper()

    if response not in ["ACCEPTED", "REJECTED"]:
        print("Response must be ACCEPTED or REJECTED.")
        return False

    indexes = coordination_db.index[
        coordination_db["offer_id"].astype(str) == str(offer_id)
    ].tolist()

    if not indexes:
        print("ERROR: Offer not found.")
        return False

    idx = indexes[-1]

    if str(coordination_db.at[idx, "status"]).upper() != "PENDING":
        print("This offer is not pending.")
        return False

    coordination_db.at[idx, "status"] = response
    coordination_db.at[idx, "responded_at"] = (
        datetime.now(timezone.utc).isoformat()
    )

    print("Offer response recorded:", response)
    return True


def find_next_ngo(food_id, rejected_ngo):
    """Create a new offer for an eligible, uncontacted NGO."""
    eligible = match_food_to_ngos(food_id)

    if eligible.empty:
        print("No eligible NGOs found.")
        return None

    previous = coordination_db[
        coordination_db["food_id"].astype(str) == str(food_id)
    ]

    already_contacted = set(previous["ngo_name"].astype(str))
    already_contacted.add(str(rejected_ngo))

    for ngo_name in eligible["ngo"].astype(str):
        if ngo_name not in already_contacted:
            print("Reassigning donation to:", ngo_name)
            return create_ngo_offer(food_id, ngo_name)

    print("No additional eligible NGO is available.")
    return None


# ================================================================
# STEP 3: RUN THE DONATION MATCHING WORKFLOW
# ================================================================

print("\nSTEP 3: DATABASE SUMMARY")

print("Donations:", len(donations_db))
print("NGOs:", len(ngos))
print("Existing offers:", len(coordination_db))
```
